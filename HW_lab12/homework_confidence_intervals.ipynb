## Exercitiul 9.7 a)

### Datele problemei

$$
n=100
$$

$$
\bar{x}=37.7
$$

$$
\sigma=9.2
$$

Se cere intervalul de incredere de 90% pentru media populatiei.

---

### Formula

Pentru abatere standard cunoscuta folosim:

$$
\bar{x}\pm z_{\alpha/2}\cdot \frac{\sigma}{\sqrt{n}}
$$

Pentru:

$$
90\%
$$

avem:

$$
z_{\alpha/2}=1.645
$$

---

### Calcul

$$
\frac{\sigma}{\sqrt{n}}=
\frac{9.2}{\sqrt{100}}
=
\frac{9.2}{10}
=
0.92
$$

$$
37.7\pm1.645\cdot0.92
$$

$$
37.7\pm1.513
$$

---

### Rezultat

$$
CI_{90\%}=(36.19,\ 39.21)
$$

---

### Concluzie

Cu incredere de 90%, media reala a numarului de utilizatori concurenti se afla intre:

$$
36.19
$$

si

$$
39.21
$$

## Exercitiul 2 - Interval de incredere pentru media unui dataset

### Datele problemei

Avem datasetul:

$$
10,\ 19,\ 11,\ 12,\ 15,\ 19,\ 9,\ 17,\ 1,\ 22,\ 9,\ 8
$$

Se cere intervalul de incredere de 95% pentru media populatiei.

---

### Date cunoscute

$$
n=12
$$

Calculam media:

$$
\bar{x}=
\frac{10+19+11+12+15+19+9+17+1+22+9+8}{12}
$$

$$
\bar{x}=
\frac{152}{12}
=
12.67
$$

---

### Formula

Deoarece abaterea standard a populatiei este necunoscuta, folosim distributia Student t:

$$
\bar{x}\pm t_{\alpha/2,n-1}\cdot SE
$$

unde:

$$
SE=\frac{s}{\sqrt{n}}
$$

---

### Calculul abaterii standard

Abaterea standard a esantionului este:

$$
s \approx 5.91
$$

Prin urmare:

$$
SE=\frac{5.91}{\sqrt{12}}
$$

$$
SE\approx1.71
$$

---

### Valoarea critica

Pentru un interval de incredere de 95%:

$$
\alpha=0.05
$$

$$
df=n-1=11
$$

Din tabelul Student:

$$
t_{\alpha/2,11}=t_{0.025,11}\approx2.201
$$

---

### Calculul intervalului

$$
12.67\pm2.201\cdot1.71
$$

$$
12.67\pm3.76
$$

---

### Rezultat

$$
CI_{95\%}=(8.91,\ 16.43)
$$

---

### Concluzie

Cu incredere de 95%, media populatiei se afla intre:

$$
8.91
$$

si

$$
16.43
$$

```python
from scipy.stats import sem, t
import numpy as np

confidence = 0.95
data = [10, 19, 11, 12, 15, 19, 9, 17, 1, 22, 9, 8]

n = len(data)
m = np.mean(data)
std_err = sem(data)

h = std_err * t.ppf((1 + confidence) / 2, n - 1)

lower = m - h
upper = m + h

print(f"n = {n}")
print(f"mean = {m:.2f}")
print(f"standard error = {std_err:.2f}")
print(f"confidence interval = ({lower:.2f}, {upper:.2f})")
```
