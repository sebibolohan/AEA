The dataset contains three groups:

- Group 1 = Control Group
- Group 2 = Mnemonic A
- Group 3 = Mnemonic B

The score represents the number of words recalled.

We want to determine whether the mean scores of the three groups are significantly different.

H0: μ1 = μ2 = μ3

The means of all groups are equal.

H1:

At least one group mean differs from the others.

DATA

```python
group1 = [3,5,3,2,4,6,9,3,8,10]

group2 = [10,8,15,9,11,16,17,15,7,10]

group3 = [29,15,14,15,17,10,8,11,18,19]
```

DESCRIPTIVE STATISTICS

```python
import numpy as np

print("Group 1 mean:", np.mean(group1))
print("Group 2 mean:", np.mean(group2))
print("Group 3 mean:", np.mean(group3))
```

ONE WAY ANOVA

```python
from scipy.stats import f_oneway

F, p = f_oneway(group1, group2, group3)

print("F statistic =", F)
print("p-value =", p)
alpha = 0.05

if p < alpha:
    print("Reject H0")
else:
    print("Fail to reject H0")
```

If p < 0.05, the mean scores are significantly different.

TURKEY POST-HOC


```python
import numpy as np
from statsmodels.stats.multicomp import pairwise_tukeyhsd

values = np.array(
    group1 +
    group2 +
    group3
)

groups = (
    ['Control'] * len(group1) +
    ['MnemonicA'] * len(group2) +
    ['MnemonicB'] * len(group3)
)

tukey = pairwise_tukeyhsd(
    values,
    groups,
    alpha=0.05
)

print(tukey)
```

The Tukey test identifies which pairs of groups differ significantly.
A pair is significantly different when reject=True.

```python
import pandas as pd
import scikit_posthocs as sp

df = pd.DataFrame({
    'group':
        ['Control']*10 +
        ['MnemonicA']*10 +
        ['MnemonicB']*10,
    'score':
        group1 + group2 + group3
})
sp.posthoc_scheffe(
    df,
    val_col='score',
    group_col='group'
)
```

Pairs with p-value < 0.05 are significantly different.
